```python
#collection: list , tuple , set ,dict
fruits = ['apple', 'muskmelon', 'strawberry', 'lichi']

print(fruits)
print(type(fruits))
print(len(fruits))
print(fruits[0])
print(fruits[-1])
print(fruits[1:3])
print(fruits[1:]) #prints from 1 to end
print(fruits[:3]) #prints from 0 to 2
print(fruits[::2]) #prints
print(fruits[::]) #prints all values
print(fruits[::-1]) #reverse


#updating element
fruits[2] = 'watermelon'

#add an element
fruits.append('blueberry')
fruits.insert(2,'Blackberry')
print(fruits)
#adding multiple values
fruits.extend(['gooseberry', 'redberry'])
print(fruits)
#remove from last
fruits.pop(0)
fruits.remove('blueberry')
print(fruits)

fruits.sort()
print(fruits)
fruits.reverse() #we can also use
fruits.sort(reverse=True)
print(fruits)
```

```python
# create a lis of cars in your garage
carss= ['Toyoto', 'Ford' , 'Hyundai', 'Thar', 'BMW']
print(carss[:])
print(carss[::-1])
carss.remove('Ford')
carss.append('Audi')
print(carss)
carss.insert(1,'Ferrari')
print(carss)
carss.extend(['Kia', 'Porche', 'Rolls-Royce'])
print("length of cars list:-", len(carss))
print("Cars are:- ", carss)
carss.sort()
print(carss)
carss.reverse()
print(carss)
```

```python
#while loop
lucky_no = 3244
while True:
  num = int(input("Enter lucky no:- "))
  if num >= 1000 and num <= 9999:
    if num == lucky_no:
      print("you won")
      break
    else:
      print("try again")

  else:
    print("enter 4 digit num")
```

```python
#take two inputs from the user id the products is greater than 5000 orint "pass" else keep tryung

while True:
  a = int(input("Enter first num :- "))
  b = int(input("Enter second num :- "))
  if a >0 and b > 0:
    if a*b > 5000:
      print("pass")
      break
    else:
      print("keep trying")
  else:
    print("Enter any positive number.")
```
